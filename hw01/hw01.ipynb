````python
# Данные: заказы за неделю (не менять!)
cities  = [" almaty", "Astana", "SHYMKENT", "Almaty ", "karaganda", "astana", "Almaty", "Aktobe", "Astana", "shymkent", "Almaty", "Atyrau"]
amounts = ["15990", "4500", "32000", "870", "12500", "56000", "2300", "9900", "18000", "7400", "41000", "3300"]
paid    = [True, True, False, True, True, False, True, True, True, False, True, True]
days    = ["пн", "пн", "вт", "вт", "ср", "ср", "чт", "пт", "пт", "сб", "сб", "вс"]
````

## [База]
1. Приведите все города к единому виду (`"Almaty"`, `"Astana"`, …) и сохраните в новый список `cities_clean`.
2. Преобразуйте `amounts` из строк в числа: список `amounts_int`.
3. Посчитайте: общее число заказов, общую сумму, средний чек (округлить до целого).
4. Сколько заказов не оплачено и на какую сумму?
5. Выведите таблицу в формате `город | сумма | статус`, используя `zip` и f-строку.

````python
# База
#1
cities_clean = cities.copy()
print("-" * 90)
for i in range(len(cities)):
    cities_clean[i] = cities[i].strip().lower().capitalize()
print(cities_clean)
#2
amounts_int = amounts.copy()
for i in range(len(amounts)):
	amounts_int[i] = int(amounts[i].strip())
print(amounts_int)
print("-" * 90)
#3
print(f'Общее число заказов: {len(amounts_int)}')
print(f'Общая сумма: {sum(amounts_int)}')
print(f'Средний чек: {round(sum(amounts_int) / len(amounts_int))}')
print("-" * 90)

#4
not_paid_number=0
not_paid_amount=0
for i in range(len(paid)):
	if paid[i]== False:
	  not_paid_number += 1
	  not_paid_amount = not_paid_amount + amounts_int[i]

print(f'Количество не оплаченных заказов:{ not_paid_number}')
print(f'Сумма не оплаченных заказов:{not_paid_amount}')
print("-" * 90)
#5
status = paid.copy()
for i in range(len(status)):
    if status[i] == True:
        status[i] = "Оплачено"
    else:
        status[i] = "Не оплачено"
print(f'City| Amounts| Status')
for city, summ, statuss in zip(cities_clean,amounts_int,status):
	print(f"{city} | {summ} | {statuss}")
````

## [Челлендж]
10. Без словарей и библиотек постройте два списка `unique_cities` и `revenue_by_city` (оплаченная выручка по каждому городу), отсортированные по убыванию выручки.
11. Найдите **медиану** суммы заказа без `statistics` и `numpy`. Учтите чётное число элементов.
12. Выведите текстовую «гистограмму» выручки по дням: каждые 5 000 ₸ рисуются как `█`.
```
пн | ████  20490
```

13. *Подумать:* какие проблемы данных вы встретили (пробелы, регистр, числа-строки)? Опишите 2–3 предложениями в Markdown-ячейке. Именно это мы будем делать в Pandas на занятии 6.

````python
# Челлендж
#10
unique_cities= list(set(cities_clean))
print(unique_cities)
revenue_by_city = []

for city in unique_cities:
    total = 0

    for i in range(len(cities_clean)):
        if cities_clean[i] == city and paid[i] == True:
            total += amounts_int[i]

    revenue_by_city.append(total)

print(unique_cities)
print(revenue_by_city)
result = sorted(zip(revenue_by_city, unique_cities), reverse=True)

revenue_by_city = []
unique_cities = []

for revenue, city in result:
    revenue_by_city.append(revenue)
    unique_cities.append(city)

print(unique_cities)
print(revenue_by_city)

#11
amounts_sorted = amounts_int.copy()
amounts_sorted.sort()

n = len(amounts_sorted)

if n % 2 == 1:
    median = amounts_sorted[n // 2]
else:
    median = (amounts_sorted[n // 2 - 1] + amounts_sorted[n // 2]) / 2

print(f'Медиана суммы заказа: {median}')

#12
unique_days = []

for day in days:
    if day not in unique_days:
        unique_days.append(day)

revenue_by_day = []

for day in unique_days:
    total = 0

    for i in range(len(days)):
        if days[i] == day and paid[i] == True:
            total += amounts_int[i]

    revenue_by_day.append(total)

for day, revenue in zip(unique_days, revenue_by_day):
    print(f"{day} | {'█' * (revenue // 5000)} {revenue}")
````
